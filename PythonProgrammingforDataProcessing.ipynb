# Section 1. Python Functions with Control Structure

**1.1-** Function takes length and width of a rectangle as parameters. Returns area of triangle.

-Program multiplies both variables to return area after verifying values are positive.


```python
def calculate_rectangle_area(length, width):
  if length<0 or width<0:
    print("Invalid input: length and width must be positive values")
  else:
    return float(length*width)

result1=calculate_rectangle_area(6,8)
result2=calculate_rectangle_area(-4,10)

print(result1)
print(result2)
```

**1.2-** Function takes account balance and returns whether its surplus, deficit, or break-even.

-Conditionals check the value against 0, then program returns balance

```python
def classify_balance(amount):
  if amount<0:
    return "Deficit"
  elif amount>0:
    return "Surplus"
  elif amount==0:
    return "break-even"

result1=classify_balance(150)
result2=classify_balance(-75)

print(result1)
print(result2)
```

# Section 2. List Operations
**2.1-** Function that takes a list of integers and returns a list with duplicates removed

-Program converts data type to sets then back to a list, in order to remove duplicates

```python
def remove_duplicates(numbers):
  dupli_removed=set(numbers)
  return list(dupli_removed)


result1=remove_duplicates([1,2,2,3,4,4,5])
result2=remove_duplicates([15, 25, 15, 35, 35, 45])
print(result1)
print(result2)
```

**2.2-** Function takes two lists as input and returns a new list only with common elements
- Lists are converted into sets then back into a list, in order to remove duplicates


```python
def list_intersection(list1, list2):
  #converted to set then back to list
  return list(set(list1).intersection(set(list2)))

result1=list_intersection([2, 4, 6, 8, 10],[6, 8, 10, 12, 14])
print(result1)
```

# Section 3: Dictionaries
**3.1-** Function takes in a dictionary with product names and customer ratings and returns the average ratings

-Program uses dictionary methods to extract keys and values, which mathematical operators are then applied to calculate the average

```python
def average_rating(ratings):

  product=ratings.keys()
  customer_ratings=ratings.values()
  #uses sum and len to calculate avg
  avg=sum(customer_ratings)/len(customer_ratings)
  return avg

result1=average_rating({'Widget': 85, 'Gadget': 92, 'Gizmo': 78})
result2=average_rating({'Sam': 70, 'Tia': 85, 'Zed': 95})

print(result1)
print(result2)
```

**3.2-** Function takes two dictionaires as input and merges them by updating common keys and adding corresponding values

-Program uses a for loop to look through each value in dict1, and mathematical operators are then applied based off the conditions met

```python
def combine_inventory(dict1,dict2):
  for key in dict2:
    #adds values together if both values are in both dicts
    if key in dict1:
      dict1[key]+=dict2[key]
    #puts value into other dict if not
    else:
      dict1[key]=dict2[key]
  return dict1

result1 = combine_inventory({'p': 10, 'q': 20, 'r': 30}, {'q': 5, 'r': 15, 's': 25})
result2= combine_inventory({'m': 100, 'n': 200}, {'n': 300, 'o': 400})

print(result1)
print(result2)
```

# Section 4: Strings
**4.1-** Function takes string as an input and returns the amount of consonants in the string (uppercase and lowercase)

-String is first converted into lowercase, then is looped through and compared against a list of consonants to get a count.

```python
cons="qwrtypsdfghjklzxcvbnm"

def count_consonants(input_string):
  count=0
  string=input_string.lower()
  for i in range(len(string)):
    if string[i] in cons:
      count+=1
  return count

result1=count_consonants("Hello, World!")
result2=count_consonants("Python is awesome")

print(result1)
print(result2)
```

**4.2-** Function takes sentence and returns the sentence with the first letter of each word capitalized

-String method is applied to capitalize the first letter of each word

```python
def title_case_sentence(sentence):
  sentence=sentence.title()
  return sentence

result1=title_case_sentence("quick brown fox")
result2=title_case_sentence("data science is powerful")

print(result1)
print(result2)
```

# Section 5: DataFrame with Pandas
**5.1-** Function takes created DataFrame and returns the total payroll amount

-Program creates DataFrame from nested list, and then indexes into the DataFrame and uses a mathematical method to get total payroll amount

```python
import pandas

data=[
   ["X", 6, 12],
   ["Y", 4, 18],
   ["Z", 3, 25],
 ]

#creating DataFrame
columns = ['Employee', 'Hours', 'Rate']
df = pandas.DataFrame(data, columns=columns)


def total_payroll(df):
  #multiplying rates across the rows, then getting the sum
  amount=(df["Hours"]*df["Rate"]).sum()
  return amount

result1=total_payroll(df)

print(result1)
```

**5.2-** Function takes a DataDrame and a threshold of number of hours and returns a new DataFrame with only rows where hours are above the threshold

-DataFrame is indexed into, then is compared against threshold parameter to create new DataFrame

```python
def filter_by_threshold(df, threshold):
  #making new DataFrame based off if rows make the condition
  newdf=df[df["Hours"]>threshold]
  return newdf

result=filter_by_threshold(df, threshold=3)
print(result)
```
