```python
from detoxify import Detoxify
import pandas as pd

# Load once, then reuse for every sentence.
model = Detoxify("original", device="cpu")

def get_detoxify_toxicity(text):
    results = model.predict(text)
    # Extract the overall toxicity score, not an average of categories.
    return float(results["toxicity"])

score = get_detoxify_toxicity("I hate how she acts in meetings.")
print("Detoxify toxicity score:", score)

# Upload your data.csv to Colab before running this cell.
df = pd.read_csv("data.csv")
df["detoxify_score"] = df["Text"].apply(get_detoxify_toxicity)
df["detoxify_label"] = df["detoxify_score"].apply(
    lambda score: "Toxic" if score >= 0.5 else "Non-Toxic"
)
df.to_csv("resultdata.csv", index=False)
display(df)
```
